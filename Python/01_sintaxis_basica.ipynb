## 1. Indentación: los bloques sin llaves

````python
temperatura = 28

if temperatura > 25:          # el if lo veremos en el tema 04
    print("Hace calor")
    print("Bebe agua")      # sigue dentro del if
print("Esto se ejecuta siempre")  # fuera del if
````

Un bloque **no puede estar vacío**. Si necesitas un bloque que no haga nada (por ejemplo, mientras diseñas el código), usa `pass` o `...` (el objeto *Ellipsis*).

## 2. Instrucciones y líneas

- Normalmente **una instrucción por línea** y sin `;` al final.
- El `;` existe para separar varias instrucciones en la misma línea, pero PEP 8 desaconseja usarlo.
- Una instrucción puede ocupar varias líneas de dos formas:
  - **Continuación implícita** (recomendada): todo lo que va dentro de paréntesis `()` (y también de `[]` o `{}`, que veremos con las colecciones) puede partirse libremente.
  - **Continuación explícita** con `\` al final de la línea (desaconsejada; verás por qué).

````python
a = 1; b = 2; c = 3   # válido, pero poco recomendable
print(a, b, c)

total = (
    a
    + b
    + c
)
print(total)

print(
    "rojo",
    "verde",
    "azul",
)
````

### 🧩 Reto 2.1 · La barra invertida traicionera  ⭐⭐


Estas dos líneas parecen idénticas. La única diferencia es que en la segunda hay **un espacio después de la `\`**.

¿Por qué falla la segunda? ¿Por qué PEP 8 prefiere los paréntesis a la `\`?

````python
correcto   = "x = 1 + \\\n2"
incorrecto = "x = 1 + \\ \n2"   # ojo: espacio tras la barra

try:
    exec(correcto)
    print("correcto: funciona")
except SyntaxError as e:
    print(f"correcto: SyntaxError: {e.msg}")

try:
    exec(incorrecto)
    print("incorrecto: funciona")
except SyntaxError as e:
    print(f"incorrecto: SyntaxError: {e.msg}")
````

## 3. Comentarios y docstrings

- **Comentario** (`#`): el intérprete lo ignora por completo. Sirve para explicar el *por qué* del código.
- **Docstring**: una cadena literal (normalmente con triple comilla) colocada como **primera instrucción** de un módulo, función o clase. **No es un comentario**: se guarda en el atributo `__doc__` y la usan `help()`, VS Code (al pasar el ratón) y los generadores de documentación.

Las funciones (`def`) las veremos en el tema 07. Aquí solo nos fijamos en **dónde va el docstring** y cómo se consulta.

````python
def area_rectangulo(base, altura):
    """Calcula el área de un rectángulo.

    Args:
        base: longitud de la base.
        altura: longitud de la altura.

    Returns:
        El área (base * altura).
    """
    # Sin validación: asumimos valores positivos
    return base * altura


print(area_rectangulo.__doc__)
````

## 4. `print()` a fondo

`print()` es una función con varios parámetros útiles:

| Parámetro | Por defecto | Para qué sirve |
| --- | --- | --- |
| `sep` | `" "` | Separador entre los valores |
| `end` | `"\n"` | Lo que se escribe al final |
| `file` | `sys.stdout` | Dónde se escribe (por ejemplo `sys.stderr`) |
| `flush` | `False` | Forzar la escritura inmediata del búfer |

````python
import sys

print("a", "b", "c")
print("a", "b", "c", sep="")
print("a", "b", "c", sep=" -> ")

print("Cargando", end="")
print("...", end="")
print(" listo")

print("Esto es un error", file=sys.stderr)
````

Fíjate en que la última línea aparece con un fondo distinto: es la salida de error (`stderr`), un flujo diferente a la salida estándar (`stdout`).

## 5. f-strings

Las f-strings (cadenas con prefijo `f`) permiten insertar **cualquier expresión** entre llaves. Después de `:` se puede añadir una **especificación de formato**.

````python
nombre = "Ana"
edad = 34
saldo = 1234567.891
ratio = 0.4567

print(f"{nombre} tiene {edad} años")
print(f"El año que viene tendrá {edad + 1}")      # expresiones
print(f"{nombre.upper()}")                       # llamadas a métodos
print(f"Saldo: {saldo:.2f}")                     # 2 decimales
print(f"Saldo: {saldo:,.2f}")                    # separador de miles
print(f"Porcentaje: {ratio:.1%}")                # porcentaje
print(f"[{nombre:<10}]")                         # alinear a la izquierda en 10
print(f"[{nombre:>10}]")                         # a la derecha
print(f"[{nombre:^10}]")                         # centrado
print(f"[{edad:05d}]")                           # rellenar con ceros
````

Dos trucos muy útiles para depurar:

- `{variable=}` muestra el nombre **y** el valor.
- `!r` usa la representación técnica del objeto (`repr`) en lugar de la legible (`str`).

````python
from datetime import date

hoy = date(2026, 10, 13)
texto = "hola\tmundo"

print(f"{edad=}")
print(f"{edad * 2 = }")
print(f"{texto}")
print(f"{texto!r}")
print(f"{hoy:%d/%m/%Y}")          # los objetos fecha tienen su propio formato
````

## 6. Identificadores, mayúsculas y palabras reservadas

- Python **distingue mayúsculas y minúsculas**: `nombre`, `Nombre` y `NOMBRE` son tres variables distintas.

- Las **palabras reservadas** (`if`, `for`, `class`, `True`…) no se pueden usar como nombres.

````python
nombre = "minúsculas"
Nombre = "Capitalizada"
NOMBRE = "MAYÚSCULAS"
print(nombre, Nombre, NOMBRE)

año = 2026
π = 3.14159
print(año, π)
````

## 7. Estilo: PEP 8 y el Zen de Python

[PEP 8](https://peps.python.org/pep-0008/) es la guía de estilo oficial. No es obligatoria para el intérprete, pero es el estándar de facto de la comunidad.

| Elemento | Convención | Ejemplo |
| --- | --- | --- |
| Variables y funciones | `snake_case` | `precio_total`, `calcular_iva()` |
| Clases | `PascalCase` | `ClienteVip` |
| Constantes | `MAYUSCULAS` | `TASA_IVA = 0.21` |
| "Privado" por convención | prefijo `_` | `_cache` |
| Indentación | 4 espacios | |
| Longitud de línea | 79 (muchos equipos usan 88 o 100) | |
| Espacios | alrededor de `=` y operadores, no dentro de paréntesis | `x = f(a, b)` |

## ✅ Resumen

- La **indentación** es sintaxis: 4 espacios, nunca mezclar con tabuladores.
- Para partir líneas, mejor **paréntesis** que `\`.
- Cuidado con las **comas**: una que falta une cadenas y una que sobra crea tuplas.
- **Docstring ≠ comentario**: es la primera cadena literal de una función, clase o módulo y se guarda en `__doc__`.
- `print()` **muestra** pero **devuelve `None`**.
- Las **f-strings** admiten expresiones, especificaciones de formato, `=` y `!r`. Su sintaxis cambió en Python 3.12.
- No uses nombres de **built-ins** (`list`, `id`, `print`…) como variables.
